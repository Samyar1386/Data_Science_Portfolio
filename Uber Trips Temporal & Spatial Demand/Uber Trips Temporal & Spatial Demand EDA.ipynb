```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


dates = pd.date_range(start='2023-05-01', periods=1500, freq='h')
df_uber = pd.DataFrame({'Timestamp': dates})
df_uber['Hour'] = df_uber['Timestamp'].dt.hour
df_uber['DayOfWeek'] = df_uber['Timestamp'].dt.day_name()


hourly_demand = df_uber.groupby('Hour').size().reset_index(name='Trip_Count')

print("=== 🚘 ساعت‌های پیک تقاضای اسنپ/اسنپ/اوبر ===")
print(hourly_demand.sort_values(by='Trip_Count', ascending=False).head(3).to_string(index=False))


plt.figure(figsize=(12, 5))
sns.lineplot(data=hourly_demand, x='Hour', y='Trip_Count', marker='o', color='purple', linewidth=2)
plt.title('Hourly Trip Demand Dynamics (24-Hour Cycle)', fontsize=12, fontweight='bold')
plt.xlabel('Hour of Day (0 - 23)')
plt.ylabel('Total Trips')
plt.xticks(range(0, 24))
plt.tight_layout()
plt.show()
```
