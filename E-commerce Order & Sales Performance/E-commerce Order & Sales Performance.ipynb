```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(42)
regions = ['North America', 'Europe', 'Asia-Pacific', 'LATAM']

df_ecom = pd.DataFrame({
    'Region': np.random.choice(regions, 400),
    'Order_Value': np.random.uniform(10, 300, 400).round(2),
    'Shipping_Days': np.random.randint(1, 10, 400)
})


region_summary = df_ecom.groupby('Region').agg(
    total_revenue=('Order_Value', 'sum'), 
    avg_ship_time=('Shipping_Days', 'mean')
).reset_index().round(2)

print("=== 🛒 عملکرد فروش و زمان ارسال منطقه‌ای ===")
print(region_summary)


plt.figure(figsize=(10, 5))
sns.barplot(data=region_summary, x='Region', y='total_revenue', palette='Blues_d')
plt.title('Total Revenue ($) by Geographical Region', fontsize=12, fontweight='bold')
plt.xlabel('Region')
plt.ylabel('Total Revenue ($)')
plt.tight_layout()
plt.show()
```
