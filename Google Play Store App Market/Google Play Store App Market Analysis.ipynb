```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(42)
df_apps = pd.DataFrame({
    'Type': np.random.choice(['Free', 'Paid'], 300, p=[0.88, 0.12]),
    'Rating': np.random.uniform(3.0, 5.0, 300).round(1),
    'Installs': np.random.choice([1000, 10000, 100000, 1000000, 10000000], 300)
})


app_summary = df_apps.groupby('Type').agg(
    avg_rating=('Rating', 'mean'), 
    median_installs=('Installs', 'median')
).reset_index().round(2)

print("=== 📱 مقایسه برنامه‌های رایگان و پولی ===")
print(app_summary)


plt.figure(figsize=(10, 5))
sns.kdeplot(data=df_apps, x='Rating', hue='Type', fill=True, common_norm=False, palette='crest')
plt.title('App Rating Density Distribution (Free vs Paid)', fontsize=12, fontweight='bold')
plt.xlabel('Rating (1.0 to 5.0)')
plt.tight_layout()
plt.show()
```
