```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(42)
budget = np.random.uniform(5, 250, 200) # میلیون دلار
gross = budget * np.random.uniform(0.5, 4.0, 200)

df_movies = pd.DataFrame({
    'Budget_M': budget.round(2), 
    'Gross_M': gross.round(2)
})
df_movies['ROI (%)'] = ((df_movies['Gross_M'] - df_movies['Budget_M']) / df_movies['Budget_M']) * 100

print("=== 🎥 ۵ فیلم برتر بر اساس درآمد کل ===")
print(df_movies.sort_values(by='Gross_M', ascending=False).head().to_string(index=False))


plt.figure(figsize=(10, 6))
sns.regplot(data=df_movies, x='Budget_M', y='Gross_M', scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.title('Movie Budget vs. Box Office Gross Revenue ($M)', fontsize=12, fontweight='bold')
plt.xlabel('Budget ($ Millions)')
plt.ylabel('Gross Revenue ($ Millions)')
plt.tight_layout()
plt.show()
```
