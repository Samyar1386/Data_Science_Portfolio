```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)
dates = pd.date_range(start='2023-01-01', periods=365, freq='D')
temp = 20 + 15 * np.sin(np.linspace(0, 2*np.pi, 365)) + np.random.normal(0, 3, 365)

df_weather = pd.DataFrame({'Date': dates, 'Temp_C': temp.round(1)})


df_weather['7D_Rolling_Avg'] = df_weather['Temp_C'].rolling(7).mean().round(1)

print("=== 🌡️ ۵ روز اول داده‌های آب و هوا ===")
print(df_weather.dropna().head().to_string(index=False))


plt.figure(figsize=(14, 5))
plt.plot(df_weather['Date'], df_weather['Temp_C'], alpha=0.4, color='gray', label='Daily Temp (°C)')
plt.plot(df_weather['Date'], df_weather['7D_Rolling_Avg'], color='crimson', linewidth=2, label='7-Day Rolling Avg')
plt.title('Annual Daily Temperature Trend & Rolling Average (2023)', fontsize=12, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```
