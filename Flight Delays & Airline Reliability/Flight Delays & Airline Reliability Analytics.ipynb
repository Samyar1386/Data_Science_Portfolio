```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(42)
carriers = ['American Air', 'Delta Air', 'United Air', 'Southwest']

df_flights = pd.DataFrame({
    'Carrier': np.random.choice(carriers, 500),
    'DepDelay': np.random.normal(10, 20, 500).round(1),
    'ArrDelay': np.random.normal(12, 25, 500).round(1)
})

df_flights['IsDelayed'] = df_flights['ArrDelay'] > 15
delay_summary = df_flights.groupby('Carrier').agg(
    delay_rate=('IsDelayed', 'mean'), 
    avg_arr_delay=('ArrDelay', 'mean')
).reset_index().round(2)
delay_summary['delay_rate (%)'] = delay_summary['delay_rate'] * 100

print("=== ✈️ نرخ تاخیر شرکت‌های هواپیمایی ===")
print(delay_summary)


plt.figure(figsize=(10, 5))
sns.barplot(data=delay_summary, x='Carrier', y='delay_rate (%)', palette='Oranges_r')
plt.title('Flight Delay Rate (>15 mins) by Airline Carrier', fontsize=12, fontweight='bold')
plt.ylabel('Delay Rate (%)')
plt.tight_layout()
plt.show()
```
