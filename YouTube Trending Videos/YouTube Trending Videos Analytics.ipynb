```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(42)
cats = ['Music', 'Gaming', 'Entertainment', 'Tech', 'Education']

df_yt = pd.DataFrame({
    'category': np.random.choice(cats, 300),
    'views': np.random.randint(10000, 5000000, 300),
    'likes': np.random.randint(500, 250000, 300),
    'comments': np.random.randint(50, 20000, 300)
})


df_yt['engagement_rate (%)'] = ((df_yt['likes'] + df_yt['comments']) / df_yt['views']) * 100
summary = df_yt.groupby('category').agg(
    avg_views=('views', 'mean'), 
    avg_engagement=('engagement_rate (%)', 'mean')
).reset_index().round(2)

print("=== 📹 خلاصه عملکرد دسته‌بندی‌های یوتیوب ===")
print(summary)


plt.figure(figsize=(10, 5))
sns.barplot(data=summary, x='category', y='avg_engagement', palette='Reds_r')
plt.title('Average Engagement Rate (%) by Video Category', fontsize=12, fontweight='bold')
plt.xlabel('Category')
plt.ylabel('Engagement Rate (%)')
plt.tight_layout()
plt.show()
```
