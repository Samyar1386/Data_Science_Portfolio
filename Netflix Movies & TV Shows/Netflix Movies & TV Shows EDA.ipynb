```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(42)
data = {
    'show_id': [f's{i}' for i in range(1, 501)],
    'type': np.random.choice(['Movie', 'TV Show'], size=500, p=[0.7, 0.3]),
    'release_year': np.random.randint(2000, 2024, size=500),
    'rating': np.random.choice(['TV-MA', 'TV-14', 'TV-PG', 'R', 'PG-13'], size=500)
}
df_netflix = pd.DataFrame(data)


type_counts = df_netflix['type'].value_counts()
print("=== 🎬 تفکیک نوع محتوای نتفلیکس ===")
print(type_counts)


fig, axes = plt.subplots(1, 2, figsize=(14, 5))


axes[0].pie(type_counts, labels=type_counts.index, autopct='%1.1f%%', colors=['#e50914', '#221f1f'], startangle=90)
axes[0].set_title('Netflix Content Ratio (Movies vs TV Shows)', fontsize=12, fontweight='bold')


sns.histplot(ax=axes[1], data=df_netflix, x='release_year', hue='type', multiple='stack', bins=20, palette=['#e50914', '#333333'])
axes[1].set_title('Content Release Volume by Year', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```
