```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(42)
boroughs = ['Manhattan', 'Brooklyn', 'Queens', 'Bronx', 'Staten Island']
room_types = ['Entire home/apt', 'Private room', 'Shared room']

df_airbnb = pd.DataFrame({
    'neighbourhood_group': np.random.choice(boroughs, 500, p=[0.4, 0.3, 0.2, 0.07, 0.03]),
    'room_type': np.random.choice(room_types, 500, p=[0.5, 0.45, 0.05]),
    'price': np.random.randint(40, 500, 500)
})


price_by_borough = df_airbnb.groupby('neighbourhood_group')['price'].agg(['mean', 'median', 'std']).round(2)
print("=== 🏙️ آمار قیمت بر حسب محله ($) ===")
print(price_by_borough)


plt.figure(figsize=(12, 6))
sns.boxplot(data=df_airbnb, x='neighbourhood_group', y='price', hue='room_type', palette='Set3')
plt.title('Airbnb Price Distribution Across NYC Boroughs & Room Types', fontsize=12, fontweight='bold')
plt.xlabel('Borough')
plt.ylabel('Price per Night ($)')
plt.legend(title='Room Type')
plt.tight_layout()
plt.show()
```
