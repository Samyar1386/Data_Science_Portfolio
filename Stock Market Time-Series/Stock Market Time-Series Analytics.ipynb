```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


np.random.seed(42)
dates = pd.date_range(start='2023-01-01', periods=250, freq='B')
price = 150 + np.cumsum(np.random.normal(0, 2, size=250))

df_stock = pd.DataFrame({'Date': dates, 'Close': price})


df_stock['SMA_20'] = df_stock['Close'].rolling(window=20).mean()
df_stock['SMA_50'] = df_stock['Close'].rolling(window=50).mean()
df_stock['Daily_Return (%)'] = df_stock['Close'].pct_change() * 100

print("=== 📈 ۵ روز پایانی داده‌های سهام ===")
print(df_stock.tail().round(2))


plt.figure(figsize=(14, 6))
plt.plot(df_stock['Date'], df_stock['Close'], label='Closing Price', color='black', alpha=0.6)
plt.plot(df_stock['Date'], df_stock['SMA_20'], label='20-Day SMA', color='blue', linewidth=1.5)
plt.plot(df_stock['Date'], df_stock['SMA_50'], label='50-Day SMA', color='orange', linewidth=1.5)

plt.title('Stock Price Trend & Moving Averages (2023)', fontsize=14, fontweight='bold')
plt.xlabel('Date')
plt.ylabel('Price ($)')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
